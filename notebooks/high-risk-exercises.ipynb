## Exercise 1: Solving the Empirical Distribution

For the empirical model, we assign a distinct parameter to each feature subgroup:
* **Parameters:** $\theta = (p_{1|00}, p_{1|01}, p_{1|10}, p_{1|11})$
* **Model:** $\hat{p}_\theta(1 | x_0, x_1) = p_{1|x_0 x_1}$

Let $S_{ij}$ be the set of $N_{ij}$ patients with features $x_0=i$ and $x_1=j$. The global negative log-likelihood decouples into independent sums for each subgroup:

$$
\ell_{ij}(p_{1|ij}) = - \sum_{n \in S_{ij}} \Big[ y_n \log(p_{1|ij}) + (1-y_n) \log(1 - p_{1|ij}) \Big]
$$

Show analytically that MLE is solved by

$$
\hat{p}_{1|ij} = \frac{1}{N_{ij}} \sum_{n \in S_{ij}} y_n
$$

### Solution

Write $p \equiv p_{1|ij}$, $N \equiv N_{ij}$, $k \equiv \sum_{n \in S_{ij}} y_n$ (the count of $y=1$ in the subgroup). Then

$$
\ell_{ij}(p) = -\Big[k\log p + (N - k)\log(1-p)\Big].
$$

Differentiate and set to zero:

$$
\frac{d\ell_{ij}}{dp} = -\frac{k}{p} + \frac{N-k}{1-p} = \frac{Np - k}{p(1-p)} = 0
\;\Longrightarrow\; \hat p_{1|ij} = \frac{k}{N} = \frac{1}{N_{ij}}\sum_{n\in S_{ij}} y_n.
$$

The second derivative $\frac{d^2\ell_{ij}}{dp^2} = \frac{k}{p^2} + \frac{N-k}{(1-p)^2} > 0$ on $(0,1)$, so this critical point is the unique minimum of the NLL — i.e. the MLE. Intuition: the MLE is just the empirical fraction of positives in each subgroup.

**Boundary cases.** If $k = 0$ the MLE is $\hat p = 0$, and if $k = N$ it is $\hat p = 1$. The formula $\hat p_{1|ij} = k/N$ still gives the right answer with the convention $0 \log 0 := 0$, which makes the NLL finite (and zero) at those boundary minimizers.

**What this cell does.** Apply the Ex 1 result $\hat p_{1|ij} = \frac{1}{N_{ij}}\sum_{n \in S_{ij}} y_n$ directly to the data. We group rows by $(x_0, x_1) = (\texttt{skin\_cancer}, \texttt{depression})$ and take the mean of `big_claim` in each group. Because `big_claim` $\in \{0,1\}$, that mean *is* (# big claims) / (# claims) — exactly the MLE formula. `empirical_mle()` wraps it so we can reuse the same formula in Ex 3 and Ex 4.

| code | math |
|---|---|
| `.groupby(['skin_cancer','depression'])` | partition records into the four subgroups $S_{ij}$ |
| `['big_claim'].agg(p_hat='mean')` | $\hat p_{1\|ij} = \frac{1}{N_{ij}}\sum_{n\in S_{ij}} y_n$ |
| `.agg(n='size')` | $N_{ij} = \lvert S_{ij}\rvert$ |

```python
# Ex 2: by the proof above, the MLE for each subgroup is just the sample mean of big_claim.
def empirical_mle(df):
    return (
        df.groupby(['skin_cancer', 'depression'])['big_claim']
          .agg(p_hat='mean', n='size')
          .reset_index()
    )

mle_full = empirical_mle(big_claim_events_df)
mle_full
```

### Interpretation of MLE Results

The estimated probabilities of a big claim increase across all sub-populations:

- Individuals with *no conditions (0,0)* have the lowest probability (~0.763).
- Having *either skin cancer (1,0)* or *depression (0,1)* increases the probability to ~0.82.
- Individuals with *both conditions (1,1)* have the highest probability (~0.833).

This suggests that both conditions are associated with a higher likelihood of a big claim, and there is a mild cumulative effect when both are present.

However, the number of observations differs substantially across groups. The (0,0) group has a much larger sample size (1563), while the (1,1) group is relatively small (114). As a result, the estimate for the (1,1) group may be less reliable and more sensitive to random variation.

Overall, the empirical MLE approach provides a simple and interpretable estimate of risk, but caution is needed when interpreting results from smaller sub-populations.

## Train, validation and test sets for MLE, empirical distribution (*)

Since there are no hyperparameters to choose for the empirical distribution model, we combine train and validation datasets.

Calculate the empirical MLE model for the train-validation set of records from 2022 and 2023, and compare to a test set of records from 2024.

### Solution

**What this cell does.** Apply the same MLE formula twice — once to the **train-val** rows (years 2022 + 2023) and once to the **test** rows (year 2024) — then put the two side-by-side per subgroup. Concretely:

$$
\hat p^{\text{train}}_{1|ij} = \frac{1}{N^{\text{train}}_{ij}} \sum_{n \in S^{\text{train}}_{ij}} y_n
\qquad\text{vs.}\qquad
\hat p^{\text{test}}_{1|ij} = \frac{1}{N^{\text{test}}_{ij}} \sum_{n \in S^{\text{test}}_{ij}} y_n
$$

`abs_diff` $= \lvert \hat p^{\text{test}}_{1|ij} - \hat p^{\text{train}}_{1|ij}\rvert$ quantifies how far the test estimate moved from the train-val one. If the relationship $p(y=1 \mid x)$ is stable over time, these diffs should be small — and smaller for subgroups with larger $N_{ij}$, since bigger samples give tighter estimates.

```python
# Ex 3: fit MLE on 2022+2023 (train-val), then compare to 2024 (held-out test).
train_val = big_claim_events_df[big_claim_events_df.year.isin([2022, 2023])]
test      = big_claim_events_df[big_claim_events_df.year == 2024]

mle_train = empirical_mle(train_val).rename(columns={'p_hat': 'p_hat_train', 'n': 'n_train'})
mle_test  = empirical_mle(test).rename(columns={'p_hat': 'p_hat_test',  'n': 'n_test'})

comparison = mle_train.merge(mle_test, on=['skin_cancer', 'depression'])
comparison['abs_diff'] = (comparison['p_hat_test'] - comparison['p_hat_train']).abs()
comparison
```

## Bootstrap statistics on the train-validation set for MLE, empirical distribution (**)

Use [empirical bootstrapping](https://en.wikipedia.org/wiki/Bootstrapping_(statistics)) to estimate the mean MLE parameters and their standard deviations on the train-validation set from above.

### Solution

**What this cell does.** Estimate how uncertain each $\hat p_{1|ij}$ from the train-val set is, by **empirical bootstrap**. For each subgroup $S^{\text{train}}_{ij}$ of size $N_{ij}$ we:

1. Draw $B = 1000$ resamples $\{y^{*(b)}_n\}_{n=1}^{N_{ij}}$ of size $N_{ij}$, **with replacement**, from the observed $\{y_n\}_{n \in S^{\text{train}}_{ij}}$.
2. Recompute the MLE on each resample:
   $$
   \hat p^{*(b)}_{1|ij} = \frac{1}{N_{ij}} \sum_{n=1}^{N_{ij}} y^{*(b)}_n \quad\text{for } b = 1, \ldots, B.
   $$
   (Same formula as Ex 1, just on the resample.)
3. Report the mean and standard deviation of those $B$ values.

`boot_mean` should match the original plug-in MLE `p_hat`. `boot_std` is a Monte-Carlo estimate of $\operatorname{se}(\hat p_{1|ij}) \approx \sqrt{\hat p_{1|ij}(1 - \hat p_{1|ij}) / N_{ij}}$ — the textbook standard error for a binomial proportion. It tells us which Ex 3 `abs_diff`s are within normal sampling noise vs. genuinely large.

```python
# Ex 4: empirical bootstrap on the train-val set. For each subgroup, draw B resamples of size N
# (with replacement) and compute the MLE on each. Report bootstrap mean and std per subgroup.
import numpy as np

rng = np.random.default_rng(seed=0)
B = 1000

rows = []
for (sc, dep), grp in train_val.groupby(['skin_cancer', 'depression']):
    y = grp['big_claim'].to_numpy()
    n = len(y)
    # Vectorised resampling: B x n matrix of bootstrap indices.
    idx = rng.integers(0, n, size=(B, n))
    boot_p_hat = y[idx].mean(axis=1)
    rows.append({
        'skin_cancer': sc, 'depression': dep, 'n': n,
        'p_hat': y.mean(),
        'boot_mean': boot_p_hat.mean(),
        'boot_std': boot_p_hat.std(ddof=1),
    })

bootstrap_df = pd.DataFrame(rows)
print(f'B = {B} bootstrap resamples per subgroup')
bootstrap_df
```

## Analytical: Decision tree number of parameters (*)

Show that the number of parameters of a decision tree for our binary random variable $(\mathbf{X},Y) = (X_0, X_1, Y)$ of big-claims prediction cannot exceed the number of parameters of the empirical distribution model.

### Solution

**What we mean by "parameter".** A *parameter* here is a predicted probability at a leaf — the only real-valued quantity a tree on binary features actually has to learn. Split thresholds on a binary feature are not free parameters: the only meaningful split is "is the feature 0 or 1?", so internal nodes carry no continuous parameter, only a feature choice.

Both features are binary, so the input space $\{0,1\} \times \{0,1\}$ has only $2 \cdot 2 = 4$ distinct points. A decision tree partitions the input space into disjoint leaves, and each leaf carries exactly **one parameter** — the predicted probability $\hat p(Y=1 \mid \text{leaf})$.

Because every leaf must contain at least one distinct input point (otherwise the split is degenerate), the number of leaves is bounded by the number of distinct input points:

$$
\#\text{leaves} \;\leq\; \lvert \{0,1\} \times \{0,1\} \rvert \;=\; 4.
$$

Hence

$$
\#\text{parameters of the tree} \;=\; \#\text{leaves} \;\leq\; 4 \;=\; \#\text{parameters of the empirical-distribution model}.
$$

Equality holds only when the tree is fully grown — i.e. it has a distinct leaf for each $(x_0, x_1)$ combination, at which point the two models are equivalent. Any pruned tree strictly reduces the parameter count by merging neighbouring cells of the input space.

## Exercise: reflecting on train / validate vs test split (*)

1. What assumptions are we making with this train / test / validate split methodology?
1. What are potential weaknesses of this assumption?
1. Come up with at least one other split logic (hint: in the above )

### Solution

We're using 2022+2023 as train-val and 2024 as test. That's basically assuming the world doesn't change much year to year — whatever pattern between features and the target existed in 2022/23 still holds in 2024.

A few things that could go wrong with this. The biggest one I noticed in the data is that the same `customer_id` shows up in multiple years — look at `X_y_raw_df.head()`, customer 0 has rows for 2022, 2023 and 2024. So we're not really testing on "new" people, just on the same people's claims in a later year. If we wanted to predict for genuinely new customers, this split wouldn't tell us how well we're doing.

Also if anything changes between years — new regulations, a pandemic effect tapering off, customers ageing into different risk bands — the 2024 distribution will look different from 2022/23 and the model will struggle. And we only get one shot at the test (one year), so the score we get is a bit luck-of-the-year.

Another way to split: hold out 20% of `customer_id`s instead of a whole year. Then train and test customers are completely disjoint and we're actually testing generalisation to people the model has never seen. Or rolling-origin CV — train on 2022, validate on 2023, then train on 2022+23, validate on 2024 — which would let us watch the drift year by year if there is any.

## Exercise: train / validation / test split methodology (*)

What are reasons for and against splitting data into train / validation / tests before doing feature selection and engineering?

### Solution

The case for splitting first is mainly about leakage. If I compute a mean, scale a column, or one-hot encode using the full dataset, I'm sneaking test-set information into my features, and my test score ends up too optimistic. Splitting first means everything with a `.fit()` step (StandardScaler, OneHotEncoder, target-based encoders, feature selectors, imputers...) only sees the train rows — which is what production would actually look like.

The case against is more practical. If I split first, all my feature engineering happens on a smaller sample, so the fitted statistics are noisier. If test contains a category that train doesn't, the encoder might break. And it's just more work — you have to wrap things in Pipelines so every fit step is contained inside each CV fold.

The notebook above actually does this properly — `lr_pipeline` and `dt_pipeline` wrap the preprocessor together with the model, so when `cross_val_score` runs, the StandardScaler and OneHotEncoder are refit on each fold's training portion.

Rule of thumb that's easy to remember: if a preprocessing step has a `.fit()`, it has to happen after the split. If it's stateless (log-transform, drop a constant column, cast a type), it doesn't matter.

## Set-up: Process data for classification

```python
# Define feature types based on your specifications
numerical_features = ['height', 'weight']
categorical_features = ['occupation_group_idx']  # 4 categories [0,1,2,3]
binary_features = ['gender_idx', 'skin_cancer', 'depression']  # Binary variables
feature_columns = numerical_features + categorical_features + binary_features

# Create ColumnTransformer
feature_preprocessor = ColumnTransformer(
    transformers=[
        # Scale numerical features (height, weight)
        ('num', StandardScaler(), numerical_features),
        
        # One-hot encode occupation_group_idx (4 categories)
        # Using drop='first' to avoid multicollinearity
        ('cat', OneHotEncoder(drop='first', sparse_output=False), categorical_features),
        
        # Keep binary features as-is (they're already 0/1)
        # Using 'passthrough' to keep them unchanged
        ('bin', 'passthrough', binary_features)
    ])
```

## Set-up: Fit logistic regression

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV, cross_val_score
from sklearn.metrics import classification_report, accuracy_score, balanced_accuracy_score

# Compare models via CV on train+validation only

# Model A: Logistic Regression (already has built-in CV)
lr_pipeline = Pipeline([
    ('preprocessor', feature_preprocessor),
    ('classifier', LogisticRegressionCV(random_state=42, max_iter=1000, scoring='balanced_accuracy'))
])
lr_cv_scores = cross_val_score(
    lr_pipeline, train_valid_raw_df[feature_columns], target_train,
    cv=5, scoring='balanced_accuracy'
)
# Fit the lr_pipeline on the full train+validation set
lr_pipeline.fit(train_valid_raw_df[feature_columns], target_train)
```

## Set-up: Fit decision trees

```python
# Model: Decision Tree with hyperparameter tuning
dt_pipeline = Pipeline([
    ('preprocessor', feature_preprocessor),
    ('classifier', DecisionTreeClassifier(random_state=42))
])
param_grid = {
    'classifier__max_depth': [2, 3, 5, 7, 10, None],
    'classifier__min_samples_leaf': [1, 5, 10, 20],
}
dt_grid = GridSearchCV(
    dt_pipeline, param_grid,
    cv=5, scoring='balanced_accuracy', n_jobs=-1
)
dt_grid.fit(train_valid_raw_df[feature_columns], target_train)
```

## Exercise: Evaluate fitted logistic regression cluster-assignment on sub-population categories (**)

This task is complicated by the presence of categorial and numerical features. For categorical features, we just take the (Cartesian) product of all possible values, and evaluate the model on those values.

With numerical features in the mix, we need to use some statistics.

### Solution

For each combination of the **categorical / binary** features we evaluate `lr_pipeline.predict_proba`. We hold the **numerical** features (`height`, `weight`) at the train-data mean, so the only thing varying across rows is the sub-population.

- Cartesian product: `occupation_group_idx` (4) $\times$ `gender_idx` (2) $\times$ `skin_cancer` (2) $\times$ `depression` (2) $= 32$ sub-populations.
- For each one we report $\hat p(\text{class} = c)$ for the three KMeans cluster classes.

```python
# Ex 6: evaluate fitted logistic regression on every sub-population (mean height/weight)

# 1) Fix numerical features at the train-data mean.
height_mean = train_valid_raw_df['height'].mean()
weight_mean = train_valid_raw_df['weight'].mean()

# 2) Cartesian product over categorical + binary feature levels.
label_mapping = {
    'occupation_group_idx': [0, 1, 2, 3],
    'gender_idx':           [0, 1],
    'skin_cancer':          [0, 1],
    'depression':           [0, 1],
}
combos = make_feature_combination_array(label_mapping)
combo_df = pd.DataFrame(combos, columns=list(label_mapping.keys()))

# 3) Insert the held-constant numerical features and reorder to match feature_columns.
combo_df['height'] = height_mean
combo_df['weight'] = weight_mean
combo_df = combo_df[feature_columns]

# 4) Predict class probabilities for each sub-population.
proba = lr_pipeline.predict_proba(combo_df)
proba_cols = [f'p(class={c})' for c in lr_pipeline.classes_]
result = pd.concat(
    [combo_df.reset_index(drop=True),
     pd.DataFrame(proba, columns=proba_cols)],
    axis=1,
)
result['predicted_class'] = lr_pipeline.predict(combo_df)
result
```

**Reading the table.** Each row is one sub-population at the average `height` / `weight`. The three `p(class=…)` columns give the model's belief over the KMeans cluster of `log(agg_claim_amount + 1)`. Recall from the earlier set-up cells that the cluster index ordering is *reversed* relative to claim size — i.e. lower class index ≈ higher claims.

**Caveats.** Holding `height` and `weight` at the mean is the simplest baseline but discards real variation: if the LR has a strong height/weight effect, a single mean can hide it. A natural follow-up is to repeat at the 25th / 75th percentiles (or marginalize over the empirical distribution of `height, weight`) and check whether `predicted_class` flips for any sub-population.

**Headline observation.** In the executed output above, `predicted_class == 0` for *every* sub-population. That is not a bug — it reflects the heavy class imbalance from the KMeans target (1590 / 48 / 34 in classes 0/1/2, as shown a few cells up). At the population-mean `height` and `weight`, none of the categorical / binary axes can move the model off class 0; the smallest `p(class=0)` in the table is still ~0.90.

**Note on the reference category.** Because `feature_preprocessor` uses `OneHotEncoder(drop='first')`, `occupation_group_idx = 0` is the implicit reference; rows with `occupation_group_idx ∈ {1, 2, 3}` should be read as shifts relative to that baseline.

## Wrapping-up: Model selection based on train + validation data split

```python
print("=== Model Selection (CV on train+validation) ===")
print(f"Logistic Regression - mean balanced accuracy: {lr_cv_scores.mean():.4f} (+/- {lr_cv_scores.std():.4f})")
print(f"Decision Tree       - mean balanced accuracy: {dt_grid.best_score_:.4f}")
print(f"  Best DT params: {dt_grid.best_params_}")

# Select model
if lr_cv_scores.mean() >= dt_grid.best_score_:
    print("\n>>> Selected model: Logistic Regression")
    final_pipeline = lr_pipeline
    final_pipeline.fit(train_valid_raw_df[feature_columns], target_train)
else:
    print("\n>>> Selected model: Decision Tree")
    final_pipeline = dt_grid.best_estimator_  # already refit on full train+validation
```
